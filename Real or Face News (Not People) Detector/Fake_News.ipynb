```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
import re
import string
import nltk
import spacy
```

```python
df_fake=pd.read_csv("Fake.csv")
df_true=pd.read_csv("True.csv")
```

```python
df_fake.head(5)
```

```python
df_true.head(5)
```

```python
df_fake.shape,df_true.shape
```

```python
df_true['class']=1
df_fake['class']=0
```

```python
df_fake.head(10)
```

```python
df_true.head(10)
```

```python
df_merged=pd.concat([df_fake,df_true],axis=0)
df_merged.head(10)
```

```python
df_merged.columns
```

```python
df=df_merged.drop(['title','subject','date'],axis=1)
df.head()
```

```python
df.isna().sum()
```

```python
df=df.sample(frac=1)
df.head()
```

```python
df.reset_index(inplace=True)
df.drop(["index"], axis = 1, inplace = True)
```

```python
df.head()
```

```python
from nltk_utils import *
```

```python
def wordopt(text):
    if text is None:
        return ""
    text=upper_to_lower(text)
    text=remove_punctuation(text)
    text=remove_stopwords(text)
    text=remove_freqwords(text)
    text=remove_rarewords(text)
    text=lematization(text)
    text=remove_emoji(text)
    text=remove_urls(text)
    text=remove_html(text)
    return text

df['text']=df['text'].apply(lambda x: wordopt(x))
df.head(5)
```

```python
x=df['text']
y=df['class']
```

```python
x.size,y.size
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer=TfidfVectorizer()
xv_train=vectorizer.fit_transform(x_train)
xv_test=vectorizer.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
```

```python
model_lr=lr.fit(xv_train,y_train)
```

```python
train_pred_lr=model_lr.predict(xv_train)
print(f"Training Score Accuracy for Logistic Regression Model : ",accuracy_score(train_pred_lr,y_train))
```

```python
test_pred_lr=model_lr.predict(xv_test)
print(f"Training Score Accuracy for Logistic Regression Model : ",accuracy_score(test_pred_lr,y_test))
```

```python
conf_lr=confusion_matrix(y_test,test_pred_lr)
print(conf_lr)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, test_pred_lr))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
model_dt=dt.fit(xv_train,y_train)
```

```python
train_pred_dt=model_dt.predict(xv_train)
print(f"Training Score Accuracy for Logistic Regression Model : ",accuracy_score(train_pred_dt,y_train))
```

```python
test_pred_dt=model_dt.predict(xv_test)
print(f"Testing Score Accuracy for Logistic Regression Model : ",accuracy_score(test_pred_dt,y_test))
```

```python
conf_dt=confusion_matrix(y_test,test_pred_dt)
print(conf_dt)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, test_pred_dt))
```

```python
from sklearn.ensemble import GradientBoostingClassifier
gb=GradientBoostingClassifier(random_state=0)
```

```python
model_gb=gb.fit(xv_train,y_train)
```

```python
train_pred_gb=model_dt.predict(xv_train)
print(f"Training Score Accuracy for Logistic Regression Model : ",accuracy_score(train_pred_gb,y_train))
```

```python
test_pred_gb=model_dt.predict(xv_test)
print(f"Testing Score Accuracy for Logistic Regression Model : ",accuracy_score(test_pred_gb,y_test))
```

```python
conf_gb=confusion_matrix(y_test,test_pred_gb)
print(conf_gb)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, test_pred_gb))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(random_state=0)
model_rf=rf.fit(xv_train,y_train)
```

```python
train_pred_rf=model_rf.predict(xv_train)
print(f"Training Score Accuracy for Logistic Regression Model : ",accuracy_score(train_pred_rf,y_train))
```

```python
test_pred_rf=model_rf.predict(xv_test)
print(f"Testing Score Accuracy for Logistic Regression Model : ",accuracy_score(test_pred_rf,y_test))
```

```python
conf_rf=confusion_matrix(y_test,test_pred_rf)
print(conf_rf)
```

```python
print(classification_report(y_test, test_pred_rf))
```

```python
def output_lable(n):
    if n == 0:
        return "Fake News"
    elif n == 1:
        return "Not A Fake News"
    
def manual_testing(news):
    testing_news = {"text":[news]}
    new_def_test = pd.DataFrame(testing_news)
    new_def_test["text"] = new_def_test["text"].apply(wordopt) 
    new_x_test = new_def_test["text"]
    new_xv_test = vectorizer.transform(new_x_test)
    pred_lr = model_lr.predict(new_xv_test)
    pred_dt = model_dt.predict(new_xv_test)
    pred_gb = model_gb.predict(new_xv_test)
    pred_rf = model_rf.predict(new_xv_test)

    return print("\n\nLR Prediction: {} \nDT Prediction: {} \nGBC Prediction: {} \nRFC Prediction: {}".format(output_lable(pred_lr[0]),output_lable(pred_dt[0]), output_lable(pred_gb[0]), output_lable(pred_rf[0])))
```
